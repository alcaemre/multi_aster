```python
occupancy = np.zeros_like(last_second_states[0])
indices = np.arange(100)

for state in last_second_states:
    occupancy += (state == 1).astype(int)

occupancy = occupancy / 1000

# plt.scatter(indices, occupancy)
# plt.ylabel('fraction of last second spent occupied')
# plt.xlabel('index of motor')
# plt.show()


import matplotlib.colors as mcolors

norm = mcolors.Normalize(vmin=occupancy.min(), vmax=occupancy.max())

cmap = plt.cm.viridis
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
plt.title('occupancy in final second of simulation')

ax.scatter(xs, ys, zs, c=occupancy, cmap=cmap, norm=norm)

mappable = plt.cm.ScalarMappable(cmap=cmap, norm=norm)
mappable.set_array(occupancy)
fig.colorbar(mappable, ax=ax, shrink=0.5, label="Occupancy")
plt.show()
```
